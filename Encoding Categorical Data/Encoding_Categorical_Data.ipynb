````python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# 1. Initialize OneHotEncoder with infrequent category handling
ohe = OneHotEncoder(
     min_frequency=101,           # Threshold is 100, so < 101 flags infrequent items
    handle_unknown='uncommon', # Safely manages new unseen brands
    sparse_output=False         # Returns a dense array (like get_dummies) instead of sparse matrix
)

# 2. Assign OHE to your specific column(s)
transformer = ColumnTransformer(
    transformers=[
        ('brand_ohe', ohe, ['brand'])
    ],
    remainder='passthrough'     # Keeps all other columns in your dataframe intact
)

# 3. Fit and transform the data
# Returns a clean DataFrame with your OHE columns
encoded_array = transformer.fit_transform(df)
encoded_cols = transformer.get_feature_names_out()
df_encoded = pd.DataFrame(encoded_array, columns=encoded_cols)

# 4. View a random sample of 5 rows
df_encoded.sample(5)
````

````python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
````

````python
df = pd.read_csv('covid_toy.csv')
df.head()
````

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],
                                                test_size=0.2)
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

# Corrected ColumnTransformer setup
ct = ColumnTransformer(
    transformers=[  # <--- Fixed the typo here
        ('tnf1', SimpleImputer(), ['fever']),
        ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
        ('tnf3', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city']),
    ],
    remainder='passthrough'
)
````

````python
X_train_new=ct.fit_transform(X_train)
X_test_new=ct.transform(X_test)
````
